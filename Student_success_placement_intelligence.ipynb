```python
# Step 1: Import libraries
import pandas as pd

# Step 2: Load the dataset
df = pd.read_csv('college_student_placement_dataset.csv')

# Step 3: Basic look at the data
print("Shape of dataset (rows, columns):", df.shape)
df.head()
```

```python
# Step 4: Understand column names and data types
df.info()
```

```python
# Step 5: Statistical summary of numeric columns
df.describe()
```

```python
# Step 6: Check unique values in categorical columns
print(df['Internship_Experience'].unique())
print(df['Placement'].unique())
```

```python
# Step 1: Check unique values in categorical columns (confirm no typos)
print("Internship_Experience values:", df['Internship_Experience'].unique())
print("Placement values:", df['Placement'].unique())

# Step 2: Check for missing values (double-check)
print("\nMissing values per column:")
print(df.isnull().sum())

# Step 3: Check for duplicate rows
print("\nNumber of duplicate rows:", df.duplicated().sum())

# Step 4: Inspect the CGPA outlier (values above 10)
print("\nRows with CGPA above 10:")
print(df[df['CGPA'] > 10])
```

```python
# Step 5: Fix the CGPA outlier by capping at 10.0
df['CGPA'] = df['CGPA'].clip(upper=10.0)

# Step 6: Confirm the fix
print("New CGPA max:", df['CGPA'].max())
```

```python
# Step 7: Confirm data types are correct
print(df.dtypes)
```

```python
# Check how many rows have CGPA above 10
print("Rows with CGPA > 10:", (df['CGPA'] > 10).sum())

# Fix: cap CGPA at 10.0 (CGPA can't logically exceed 10)
df['CGPA'] = df['CGPA'].clip(upper=10.0)

# Confirm the fix worked
print("New CGPA max:", df['CGPA'].max())
```

```python
# ---- Step 3A: Normalize raw columns to a 0-10 scale ----

def normalize_0_10(series):
    return (series - series.min()) / (series.max() - series.min()) * 10

df['IQ_norm'] = normalize_0_10(df['IQ'])
df['CGPA_norm'] = normalize_0_10(df['CGPA'])
df['Prev_Sem_norm'] = normalize_0_10(df['Prev_Sem_Result'])
df['Academic_Perf_norm'] = normalize_0_10(df['Academic_Performance'])
df['Comm_norm'] = normalize_0_10(df['Communication_Skills'])
df['Projects_norm'] = normalize_0_10(df['Projects_Completed'])
df['ExtraCurr_norm'] = normalize_0_10(df['Extra_Curricular_Score'])
```

```python
# ---- Step 3B: Academic Score ----
df['Academic_Score'] = (
    df['CGPA_norm'] * 0.4 +
    df['Prev_Sem_norm'] * 0.3 +
    df['Academic_Perf_norm'] * 0.3
).round(2)
```

```python
# ---- Step 3C: Skill Score ----
df['Skill_Score'] = (
    df['Comm_norm'] * 0.4 +
    df['Projects_norm'] * 0.4 +
    df['ExtraCurr_norm'] * 0.2
).round(2)
```

```python
# ---- Step 3D: Placement Score ----
df['Placement_Score'] = (
    df['Academic_Score'] * 0.4 +
    df['Skill_Score'] * 0.4 +
    df['IQ_norm'] * 0.2
).round(2)
```

```python
# ---- Step 3E : Placement Readiness using percentiles ----

p25 = df['Placement_Score'].quantile(0.25)
p75 = df['Placement_Score'].quantile(0.75)

def readiness_label_v2(score):
    if score >= p75:
        return 'HIGH'
    elif score >= p25:
        return 'MEDIUM'
    else:
        return 'LOW'

df['Placement_Readiness'] = df['Placement_Score'].apply(readiness_label_v2)

print("New Placement Readiness distribution:")
print(df['Placement_Readiness'].value_counts())
```

```python
# ---- Step 3F : Risk Level using percentiles ----

r25 = df['Risk_Score'].quantile(0.25)
r75 = df['Risk_Score'].quantile(0.75)

def risk_label_v2(score):
    if score >= r75:
        return 'HIGH'
    elif score >= r25:
        return 'MEDIUM'
    else:
        return 'LOW'

df['Risk_Level'] = df['Risk_Score'].apply(risk_label_v2)

print("\nNew Risk Level distribution:")
print(df['Risk_Level'].value_counts())
```

```python
# ---- Step 3G: Skill Gap ----
IDEAL_SKILL_SCORE = 8.0  # benchmark for a "well-prepared" student
df['Skill_Gap'] = (IDEAL_SKILL_SCORE - df['Skill_Score']).clip(lower=0).round(2)
```

```python
# Step 1: Select final columns needed for Power BI
# (drop the temporary _norm helper columns, keep everything meaningful)

final_columns = [
    'College_ID', 'IQ', 'Prev_Sem_Result', 'CGPA', 'Academic_Performance',
    'Internship_Experience', 'Extra_Curricular_Score', 'Communication_Skills',
    'Projects_Completed', 'Placement',
    'Academic_Score', 'Skill_Score', 'Placement_Score', 'Placement_Readiness',
    'Risk_Score', 'Risk_Level', 'Skill_Gap'
]

df_final = df[final_columns]

# Step 2: Quick final check before export
print("Final shape:", df_final.shape)
df_final.head()
```

```python
# Step 3: Export to CSV
df_final.to_csv('cleaned_student_data.csv', index=False)

print("File exported successfully!")
```
